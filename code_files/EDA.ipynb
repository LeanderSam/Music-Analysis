# **4) Bivariate Analysis: Correlation Heatmap**

```python
numerical_cols = ['energy', 'tempo', 'danceability', 'loudness', 'liveness', 
                  'valence', 'speechiness', 'instrumentalness', 'duration_ms', 'acousticness']

plt.figure(figsize=(12, 8))
corr_matrix = df[numerical_cols].corr()

# Plotting the heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Heatmap of Audio Features', fontsize=16)
plt.tight_layout()
plt.show()
```

### Correlation Heatmap Summary

The heatmap shows relationships between audio features. Strong positive correlations are observed between **energy and loudness (0.80)**, and moderate correlations between **danceability, valence, and loudness**, suggesting these features tend to increase together.

Strong negative correlations appear between **acousticness and energy (-0.75)**, as well as **instrumentalness with energy and loudness**, indicating that more acoustic or instrumental tracks are generally less energetic and quieter.

Most other features (e.g., tempo, duration) show weak or near-zero correlations, meaning they are relatively independent.

```python
# --- Top correlates with popularity ---
corr = df[numerical_cols + ['track_popularity']].corr()
pop_corr = corr['track_popularity'].drop('track_popularity').sort_values()
fig, ax = plt.subplots(figsize=(8, 5))
pop_corr.plot(kind='barh', ax=ax, color=['tomato' if v < 0 else 'steelblue' for v in pop_corr])
ax.axvline(0, color='black', linewidth=0.8)
ax.set_title('Correlation with Track Popularity', fontweight='bold')
ax.set_xlabel('Pearson r')
plt.tight_layout()
plt.show()
```

# 5 )Feature Engineering

```python
# Features we'll cluster on — chosen because they describe the *sound* of a track,
# not its metadata (no artist, genre, album etc.)
cluster_features_raw = [
    'danceability', 'energy', 'loudness', 'acousticness',
    'valence', 'tempo', 'speechiness', 'instrumentalness',
    'liveness', 'duration_ms'
]

# Drop the single row with nulls before we do anything
df = df.dropna(subset=cluster_features_raw).reset_index(drop=True)

print('Dataset shape after dropping nulls:', df.shape)
```

```python
# Log-transform the heavily right-skewed features to pull in extreme outliers.
# Skewness check earlier showed: duration_ms (3.99), liveness (2.36),
# speechiness (2.06), instrumentalness (1.35) — all candidates for log1p.
import numpy as np

skewed_features = ['duration_ms', 'liveness', 'speechiness', 'instrumentalness']

for feat in skewed_features:
    df[f'log_{feat}'] = np.log1p(df[feat])

# Final feature list used for clustering
cluster_features = [
    'danceability', 'energy', 'loudness', 'acousticness', 'valence', 'tempo',
    'log_speechiness', 'log_instrumentalness', 'log_liveness', 'log_duration_ms'
]

print('Features going into clustering:')
for f in cluster_features:
    print(f'  {f}')
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[cluster_features])

print('Scaled matrix shape:', X_scaled.shape)
print('Mean per feature (should be ~0):', X_scaled.mean(axis=0).round(4))
print('Std  per feature (should be ~1):', X_scaled.std(axis=0).round(4))
```

# 6) Popularity Binning

```python
# Bin continuous popularity (11–100) into three interpretable tiers.
# Thresholds chosen to give roughly balanced groups based on the distribution:
#   Q1 ≈ 41  |  Q3 ≈ 72
bins   = [0,  40,  70,  100]
labels = ['Low (<40)', 'Mid (40-70)', 'High (>70)']

df['popularity_tier'] = pd.cut(
    df['track_popularity'],
    bins=bins,
    labels=labels
)

print('Songs per tier:')
print(df['popularity_tier'].value_counts().sort_index())

tier_colors = ['#e05c5c', '#f0a500', '#4caf7d']
fig, ax = plt.subplots(figsize=(7, 4))
df['popularity_tier'].value_counts().sort_index().plot(
    kind='bar', color=tier_colors, edgecolor='white', ax=ax
)
ax.set_title('Song Count by Popularity Tier', fontsize=14, fontweight='bold')
ax.set_xlabel('Popularity Tier')
ax.set_ylabel('Number of Songs')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```

# 7 ) Choosing K: Elbow method and Silhouette Scores

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

k_range = range(2, 11)
inertias, sil_scores = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_k = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels_k))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Elbow
axes[0].plot(list(k_range), inertias, marker='o', color='#1DB954', linewidth=2)
axes[0].axvline(x=4, color='gray', linestyle='--', alpha=0.6, label='Chosen K=4')
axes[0].set_title('Elbow Method — Inertia vs K', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Number of Clusters (K)')
axes[0].set_ylabel('Inertia')
axes[0].legend()

# Silhouette
axes[1].plot(list(k_range), sil_scores, marker='s', color='#764ba2', linewidth=2)
axes[1].axvline(x=4, color='gray', linestyle='--', alpha=0.6, label='Chosen K=4')
axes[1].set_title('Silhouette Score vs K', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Number of Clusters (K)')
axes[1].set_ylabel('Silhouette Score (higher = better)')
axes[1].legend()

plt.tight_layout()
plt.show()

print('Silhouette scores by K:')
for k, s in zip(k_range, sil_scores):
    print(f'  K={k}: {s:.4f}')
```

### Elbow & Silhouette Summary

The elbow curve shows the sharpest drop in inertia between K=2 and K=4, after which gains diminish. 
The silhouette score peaks at K=2 but stabilises around K=4. We choose **K=4** as the best balance: 
K=2 is too coarse to reveal meaningful musical sub-groups, while K=5+ adds clusters without a clear interpretive payoff.

# 8) K-Means Clustering & PCA Visualisation

```python
BEST_K = 4
cluster_palette = {0: '#e05c5c', 1: '#1DB954', 2: '#4a90d9', 3: '#f0a500'}

km_final = KMeans(n_clusters=BEST_K, random_state=42, n_init=10)
df['cluster'] = km_final.fit_predict(X_scaled)

print('Songs per cluster:')
print(df['cluster'].value_counts().sort_index())
```

```python
from sklearn.decomposition import PCA

# Reduce to 2D so we can visualise the clusters
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

df['pca_1'] = X_pca[:, 0]
df['pca_2'] = X_pca[:, 1]

print(f'Variance explained by PC1: {pca.explained_variance_ratio_[0]:.1%}')
print(f'Variance explained by PC2: {pca.explained_variance_ratio_[1]:.1%}')
print(f'Total variance captured:   {pca.explained_variance_ratio_.sum():.1%}')

fig, ax = plt.subplots(figsize=(10, 7))
for c in range(BEST_K):
    mask = df['cluster'] == c
    ax.scatter(
        df.loc[mask, 'pca_1'],
        df.loc[mask, 'pca_2'],
        c=cluster_palette[c],
        label=f'Cluster {c}',
        alpha=0.5, s=20, edgecolors='none'
    )

# Plot centroids projected into PCA space
centroids_pca = pca.transform(km_final.cluster_centers_)
ax.scatter(
    centroids_pca[:, 0], centroids_pca[:, 1],
    c='white', edgecolors='black', s=200, marker='*',
    zorder=5, label='Centroids'
)
for i, (x, y) in enumerate(centroids_pca):
    ax.annotate(f'C{i}', (x, y), textcoords='offset points',
                xytext=(8, 4), fontsize=11, fontweight='bold')

ax.set_title('K-Means Clusters in PCA Space (K=4)', fontsize=14, fontweight='bold')
ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%} variance)')
ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%} variance)')
ax.legend()
plt.tight_layout()
plt.show()
```

### 9) Cluster Profiling

```python
# Mean of original (un-transformed) features per cluster
profile_features = [
    'track_popularity', 'danceability', 'energy', 'loudness',
    'acousticness', 'valence', 'tempo', 'speechiness', 'instrumentalness'
]

cluster_profile = df.groupby('cluster')[profile_features].mean().round(2)
print('Cluster Profiles (mean values):')
display(cluster_profile)
```

```python
# Heatmap of cluster centroids — normalised 0-1 for readability across different scales
from sklearn.preprocessing import MinMaxScaler

display_features = ['danceability', 'energy', 'loudness', 'acousticness',
                    'valence', 'tempo', 'speechiness', 'instrumentalness']

profile_norm = cluster_profile[display_features].copy()
profile_norm = (profile_norm - profile_norm.min()) / (profile_norm.max() - profile_norm.min())

fig, ax = plt.subplots(figsize=(12, 4))
sns.heatmap(
    profile_norm,
    annot=cluster_profile[display_features],  # show original values as annotations
    fmt='.2f',
    cmap='YlOrRd',
    linewidths=0.5,
    ax=ax,
    cbar_kws={'label': 'Relative intensity (0-1 normalised)'}
)
ax.set_title('Cluster Centroid Heatmap (annotated values = original scale)',
             fontsize=13, fontweight='bold')
ax.set_xlabel('')
plt.tight_layout()
plt.show()
```

```python
# Top genres per cluster — who lives in each cluster?
fig, axes = plt.subplots(1, BEST_K, figsize=(16, 5), sharey=False)

for c in range(BEST_K):
    genre_counts = (
        df[df['cluster'] == c]['playlist_genre']
        .value_counts()
        .head(6)
    )
    genre_counts.plot(
        kind='barh', ax=axes[c],
        color=cluster_palette[c], edgecolor='white'
    )
    avg_pop = df[df['cluster'] == c]['track_popularity'].mean()
    axes[c].set_title(
        f'Cluster {c}\nAvg Popularity: {avg_pop:.1f}',
        fontsize=11, fontweight='bold'
    )
    axes[c].invert_yaxis()
    axes[c].set_xlabel('Song count')

plt.suptitle('Top Genres per Cluster', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Popularity tier distribution within each cluster (as %)
tier_dist = (
    df.groupby(['cluster', 'popularity_tier'])
    .size()
    .unstack(fill_value=0)
    .apply(lambda r: r / r.sum() * 100, axis=1)
)

tier_dist.plot(
    kind='bar',
    figsize=(9, 5),
    color=['#e05c5c', '#f0a500', '#4caf7d'],
    edgecolor='white',
    width=0.7
)
plt.title('Popularity Tier Mix per Cluster (%)', fontsize=13, fontweight='bold')
plt.xlabel('Cluster')
plt.ylabel('% of Songs')
plt.xticks(rotation=0)
plt.legend(title='Popularity Tier', bbox_to_anchor=(1.01, 1))
plt.tight_layout()
plt.show()
```

## 10) Feature Importance via Random

```python
# Train a Random Forest to predict cluster labels — its feature importances reveal
# which audio features are doing the most work in separating the clusters.
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_scaled, df['cluster'])

importances = pd.Series(
    rf.feature_importances_,
    index=cluster_features
).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(8, 6))
importances.plot(
    kind='barh', color='#1DB954', edgecolor='white', ax=ax
)
ax.set_title(
    'Feature Importance for Cluster Separation\n(Surrogate Random Forest)',
    fontsize=13, fontweight='bold'
)
ax.set_xlabel('Importance Score')
ax.axvline(x=importances.mean(), color='gray', linestyle='--',
           alpha=0.7, label=f'Mean ({importances.mean():.3f})')
ax.legend()
plt.tight_layout()
plt.show()

print('Top 5 features driving cluster separation:')
for feat, imp in importances.sort_values(ascending=False).head(5).items():
    print(f'  {feat:<28} {imp:.4f}')
```
