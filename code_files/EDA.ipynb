```python
df = pd.read_csv('../data/combined_spotify_data.csv')
display(df.head())
display(df.tail())
```

# **2) Data Cleaning**

```python
#Identify and quantify missing values in each column
missing_values = df.isna().sum()
print("Missing values per column:\n", missing_values[missing_values > 0])

# Decide on an imputation strategy or drop rows with missing values
# We drop any rows that have missing values to keep the dataset simple and clean
print(f"\nOriginal shape: {df.shape}")
df.dropna(inplace=True)
print(f"Shape after dropping NaNs: {df.shape}")
```

```python
# Check for and remove exact duplicate rows (common in music datasets)
duplicates = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicates}")
if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print(f"Shape after dropping duplicates: {df.shape}")
```

```python
# Verify that categorical columns (like 'playlist_genre') don't have messy/overlapping text entries
if 'playlist_genre' in df.columns:
    print("\nUnique playlist genres:")
    print(df['playlist_genre'].unique())
    # Standardize string format (lowercase and trim whitespace)
    df['playlist_genre'] = df['playlist_genre'].str.lower().str.strip()
```

# **3) Univariate Analysis**

```python
# Set aesthetics for plots
sns.set_theme(style="whitegrid")

# Define numerical features of interest
numerical_features = ['danceability', 'energy', 'loudness', 'speechiness','acousticness', 'instrumentalness', 'liveness', 'valence','tempo', 'duration_ms']

# 1. Plot histograms with KDE
plt.figure(figsize=(14, 10))
for i, feature in enumerate(numerical_features, 1):
    plt.subplot(5, 2, i)
    sns.histplot(df[feature], kde=True, bins=30, color='skyblue')
    plt.title(f'Distribution of {feature}', fontsize=14)
    plt.xlabel(feature, fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    
plt.tight_layout()
plt.show()
```

```python
# 2. Bar chart for categorical variable (playlist_genre)
plt.figure(figsize=(12, 6))
# Get ordered categories by count
genre_order = df['playlist_genre'].value_counts().index
sns.countplot(y='playlist_genre', data=df, order=genre_order, hue='playlist_genre', legend=False, palette='viridis')
plt.title('Count of Tracks per Playlist Genre', fontsize=16)
plt.xlabel('Count', fontsize=12)
plt.ylabel('Genre', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# 3. Check for significant skewness in numerical data
skewness = df[numerical_features].skew()
print("Skewness of numerical features:\n", skewness)
print("\nNote: Values > 1 or < -1 indicate a highly skewed distribution.")
```

```python
# Set aesthetics for plots
sns.set_theme(style="whitegrid")

# Define numerical features for outlier detection (including tempo and duration_ms)
outlier_features = ['danceability', 'energy', 'loudness', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence',
    'tempo', 'duration_ms']

# Plot boxplots to visualize outliers
plt.figure(figsize=(16, 12))
for i, feature in enumerate(outlier_features, 1):
    plt.subplot(5, 2, i)
    sns.boxplot(x=df[feature], color='lightgreen')
    plt.title(f'Boxplot of {feature}', fontsize=14)
    plt.xlabel(feature, fontsize=12)
    
plt.tight_layout()
plt.show()
```

# **4) Bivariate Analysis: Correlation Heatmap**

```python
numerical_cols = ['energy', 'tempo', 'danceability', 'loudness', 'liveness', 
                  'valence', 'speechiness', 'instrumentalness', 'duration_ms', 'acousticness']

plt.figure(figsize=(12, 8))
corr_matrix = df[numerical_cols].corr()

# Plotting the heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Heatmap of Audio Features', fontsize=16)
plt.tight_layout()
plt.show()
```

### Correlation Heatmap Summary

The heatmap shows relationships between audio features. Strong positive correlations are observed between **energy and loudness (0.80)**, and moderate correlations between **danceability, valence, and loudness**, suggesting these features tend to increase together.

Strong negative correlations appear between **acousticness and energy (-0.75)**, as well as **instrumentalness with energy and loudness**, indicating that more acoustic or instrumental tracks are generally less energetic and quieter.

Most other features (e.g., tempo, duration) show weak or near-zero correlations, meaning they are relatively independent.

```python
# --- Top correlates with popularity ---
corr = df[numerical_cols + ['track_popularity']].corr()
pop_corr = corr['track_popularity'].drop('track_popularity').sort_values()
fig, ax = plt.subplots(figsize=(8, 5))
pop_corr.plot(kind='barh', ax=ax, color=['tomato' if v < 0 else 'steelblue' for v in pop_corr])
ax.axvline(0, color='black', linewidth=0.8)
ax.set_title('Correlation with Track Popularity', fontweight='bold')
ax.set_xlabel('Pearson r')
plt.tight_layout()
plt.show()
```

```python
# --- Popularity by genre (box plot) ---
fig, ax = plt.subplots(figsize=(10, 5))
order = df.groupby('playlist_genre')['track_popularity'].median().sort_values(ascending=False).index
sns.boxplot(data=df, x='playlist_genre', y='track_popularity', order=order,
            palette='Set2', ax=ax)
ax.set_title('Track Popularity by Genre', fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
